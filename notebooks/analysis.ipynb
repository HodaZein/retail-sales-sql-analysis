```python
import sqlite3
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

DB = Path('..') / 'data' / 'retail.db'
con = sqlite3.connect(DB)
```

## Monthly revenue trend

```python
monthly = pd.read_sql_query(open('../sql/analysis.sql').read().split(';')[0] + ';', con)
monthly['month'] = pd.to_datetime(monthly['month'])
ax = monthly.plot(x='month', y='revenue_gbp', figsize=(10, 4), legend=False)
ax.set_title('Monthly revenue (GBP)')
ax.set_ylabel('Revenue (GBP)')
plt.tight_layout()
```

## Country mix (excl. UK)

```python
country = pd.read_sql_query(open('../sql/analysis.sql').read().split(';')[2] + ';', con)
country.head(15).plot.barh(x='country', y='revenue_gbp', figsize=(8, 6), legend=False)
plt.gca().invert_yaxis()
plt.title('Top non-UK markets by revenue')
plt.tight_layout()
```
